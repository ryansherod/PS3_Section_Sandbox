**b)** The outcome of this study is `anyhelp` = whether or not the person received any help. 0 for no help, and 1 for any help.

> **How to interpret proportion outcomes:** Because `anyhelp` is only ever 0 or 1, its average is the share of interactions in which the person got help. For example, if the person got help in 3 out of 4 interactions, the average of `anyhelp` is (1 + 1 + 1 + 0) / 4 = 0.75, or 75%. So the intercept is a proportion, and the other estimates are differences in proportions. Multiply them by 100 to read them as percentage points: an estimate of 0.05 means a 5 percentage point higher chance of getting help.
